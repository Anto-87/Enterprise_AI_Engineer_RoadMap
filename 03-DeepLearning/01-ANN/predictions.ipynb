```python
import tensorflow as tf 
from tensorflow.keras.models import load_model
import pickle
import pandas as pd
import numpy as np
```

```python
# Load the trained model , scaler and encoder
model = load_model("models/ann_model.h5")

with open("models/label_encoder_gender.pkl","rb") as f:
    label_encoder_gender = pickle.load(f)

with open("models/onehot_encoder_geo.pkl","rb") as f:
    label_encoder_geo = pickle.load(f)

with open("models/scaler.pkl","rb") as f:
    scaler = pickle.load(f)
```

```python
input_data = {
    "CreditScore": 600,
    "Geography": "France",
    "Gender": "Male",
    "Age": 40,
    "Tenure": 3,
    "Balance": 60000,
    "NumOfProducts": 2,
    "HasCrCard": 1,
    "IsActiveMember": 1,
    "EstimatedSalary": 50000}
```

```python
# one-hot encode Geography
geo_encoded = label_encoder_geo.transform([[input_data["Geography"]]]).toarray()
geo_encoded_df =pd.DataFrame(geo_encoded,columns=label_encoder_geo.get_feature_names_out(["Geography"]))
```

```python
geo_encoded_df
```

```python
input_df = pd.DataFrame([input_data])
input_df
```

```python
input_df['Gender'] = label_encoder_gender.transform([input_data["Gender"]])
input_df
```

```python
input_data = pd.concat([pd.DataFrame(input_df.drop("Geography",axis=1)),geo_encoded_df],axis=1)
```

```python
input_data
```

```python
scaler.feature_names_in_
```

```python
# Scaling the input data 
input_scaled = scaler.transform(input_data)
input_scaled
```

```python
# prediction 
prediction = model.predict(input_scaled)
prediction
```

```python
prediction_prob = prediction[0][0]
prediction_prob
```

```python
if prediction_prob > 0.5:
    print("The customer is likely to leave the bank")
else:
    print("The customer is likely to stay with the bank")
```
