### Chroma

```python
from langchain_chroma import Chroma
from langchain_community.document_loaders import TextLoader
from langchain_community.embeddings import OllamaEmbeddings
from langchain_text_splitters import RecursiveCharacterTextSplitter
```

```python
loader=TextLoader("speech.txt")
documents=loader.load()
text_splitter=RecursiveCharacterTextSplitter(chunk_size=500,chunk_overlap=30)
docs=text_splitter.split_documents(documents)
```

```python
docs
```

```python
embeddings = OllamaEmbeddings(model="embeddinggemma:latest")
```

```python
vectordb = Chroma.from_documents(docs, embeddings)
```

```python
query="How does the speaker describe the desired outcome of the war?"
docs=vectordb.similarity_search(query)
docs
```

```python
docs[0].page_content
```

```python
vectordb_1 = Chroma.from_documents(docs, embeddings, persist_directory="./chroma_db")
```

```python
docs=vectordb_1.similarity_search(query)
print(docs[0].page_content)
```

```python
result = vectordb_1.similarity_search_with_score(query)
result
```

```python
reteriver = vectordb_1.as_retriever()
reteriver.invoke(query)[0].page_content
```
