#### Embedding Techniques
Converting text in to vectors

```python
import os
from dotenv import load_dotenv
load_dotenv()
```

```python
os.environ['OPENAI_API_KEY']=os.getenv('OPENAI_API_KEY')
```

```python
from langchain_openai import OpenAIEmbeddings
embeddings=OpenAIEmbeddings(model="text-embedding-3-large")
embeddings
```

```python
text ="This is a tutorial on Anthropic embedding techniques"

embedding = embeddings.embed_query(text)
embedding
```

```python
len(embedding)
```

```python
embeddings_1024=OpenAIEmbeddings(model="text-embedding-3-large", dimensions=1024)
embedding_1024 = embeddings_1024.embed_query(text)
embedding_1024
```

```python
len(embedding_1024)
```

```python
from langchain_community.document_loaders import TextLoader

loader = TextLoader("speech.txt")
txts=loader.load()
txts
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
text_splitter = RecursiveCharacterTextSplitter(chunk_size=500, chunk_overlap=50)
docs_chunks = text_splitter.split_documents(txts)
docs_chunks
```

```python
## Vector embedding and vector store

from langchain_community.vectorstores import Chroma
db = Chroma.from_documents(docs_chunks, embeddings_1024)
db
```

```python
query="We have borne with their present government through all these bitter "
reterived_results=db.similarity_search(query)
print(reterived_results)
```
