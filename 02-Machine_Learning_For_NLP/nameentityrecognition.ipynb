##### Name Entity Recoginition

```python
sentence="""The Eiffel Tower was built from 1887 to 1899 by French Engineer Gustave Eiffel, whose compnay specialized in 
building metal frameworks and structures"""
```

```python
import nltk 
words= nltk.word_tokenize(sentence)
```

```python
tag_elements = nltk.pos_tag(words)
```

```python
nltk.download('maxent_ne_chunker')
```
