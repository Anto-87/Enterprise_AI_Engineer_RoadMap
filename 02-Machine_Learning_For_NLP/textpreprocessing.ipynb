#### Stemming

Stemming is the process of reducing a words to its word stem that affies to suffixes and prefixes or to the roots of words known as lemma.

```python
## Classification Problem 
## Comments of product is a positive review or negative review 
## Reviews ---> eating, eat, eaten

words=["writes","writing", "programming", "programs","history", "finally","finalize"]
```

```python
## PorterStemmer
from nltk.stem import PorterStemmer
steming=PorterStemmer()
```

```python
for word in words:
    print(word+"----"+steming.stem(word))
```

```python
print(steming.stem('Congratulations'))
```

```python
# RegexpStemmer Class

from nltk.stem import RegexpStemmer
reg_stemmer = RegexpStemmer('ing$|s$|e$|able$', min=4)
```

```python
reg_stemmer.stem('sitting') #ing specified the expression that's why produced as Sitt
```

```python
reg_stemmer.stem('eating')
```

```python
# Snowball Stemmer

from nltk.stem import SnowballStemmer
snowballstemmer = SnowballStemmer('english')
```

```python
for word in words:
    print(word+"--->"+ snowballstemmer.stem(word))
```

```python
steming.stem('fairly'),steming.stem('sportingly')
```

```python
snowballstemmer.stem('fairly'),snowballstemmer.stem('sportingly') # Provide the extact stem words
```
