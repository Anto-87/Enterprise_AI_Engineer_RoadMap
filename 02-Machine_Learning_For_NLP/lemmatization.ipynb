#### Lemmatization

```python
# Q&A, Chat bot, Text summarization as uses cases 
from nltk.stem import WordNetLemmatizer 
```

```python
lemmatizer = WordNetLemmatizer()
```

```python
'''
Noun -n 
verb -v
adjective -a
adverb -r

'''
lemmatizer.lemmatize('buying', pos='n')
```

```python
lemmatizer.lemmatize('buying', pos='v')
```

```python
lemmatizer.lemmatize('buying', pos='a')
```

```python
lemmatizer.lemmatize('buying', pos='r')
```

```python
words=["writes","writing", "programming", "programs","history", "finally","finalize"]
for word in words:
    print(word+"-->"+lemmatizer.lemmatize(word, pos='n'))
```
