### Python Operators

```python
# Arithmetic operations 

a=10
b=5 
multiply=a*b
add=a+b
sub=a-b
div=a/b
mod=a%b
floor_div=a//b # Removes the decimal part and gives the output
expo=a**b 
print(multiply, add, sub, div, mod,floor_div, expo)
```

```python
# Comparison operators
a=10
b=10

result=(a==b)
print(result)
```

```python
# Logical Operators AND/OR/NOT

x= True 
y=True 

result = x and y 
print(result) 
```

```python
# Another scenario for and operator 
y= False
result = x and y
print(result) 

# not operator 
z= False
result = not z
print(result)
```
