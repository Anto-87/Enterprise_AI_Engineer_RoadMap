#### Python Conditional Statements

```python
## If and else statement

age = 8
if age>=18:
    print("Eligible for voting")
else:
    print("Not eligible for voting")
```

```python
## elif statement

amount =60

if amount<=50:
    print("Not eligible for buying any products!")
elif amount>=50 and amount<=70:
    print("Eligible to buy with condition apply products only!!")
else:
    print("You can buy anything as in the shop has")
    
```

```python
# nested conditonal statements 

num = input("Enter the number  >>")
num = int(num)

if num >0:
    print("Positive!")
    if num%2 ==0:
        print("The given number as even!")
    else:
        print("The given number as odd!")
else:
    print("Negative!")
        
    
```
