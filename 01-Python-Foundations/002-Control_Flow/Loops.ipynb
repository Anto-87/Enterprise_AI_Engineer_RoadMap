### Python Loops

```python
# For Looping
for i in range(0,5):
    print(i)
```

```python
for i in range(1,10,2):
    print(i)
```

```python
for i in range(10,1,-1):
    print(i)
```

```python
# Strings 

str = "Anto"
for i in str:
    print(i)
```

```python
# While looping
count=0
while count<5:
    print(count)
    count+=1
    
```

```python
count =0
while count%2==0:
    print(count)
    count+=1
    
```

```python
# Break
for j in range(1,10):
    if j ==5:
        break
    print(j)
```

```python
# Continue 

for i in range(1,5):
    if i%2 == 0:
        continue 
    print("The odd no's are", i)
```

```python
# pass 

for z in range(5):
    if z%2==0:
        print("Inside the pass loop")
        pass # do nothing 
    print(z)
```

```python
# Nested Loops
for i in range(3):
    for j in range(5):
        print(f"i:{i},j:{j}")
```
