```python
import csv 
with open('text.csv', mode='w',newline='') as file:
    writer =csv.writer(file)
    writer.writerow(['name','age'])
    writer.writerow(['Chris',34])
    
with open('text.csv', mode='r') as file:
    reader =csv.reader(file)
    for row in reader:
        print(row)

    
```
