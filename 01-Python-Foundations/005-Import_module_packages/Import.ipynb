#### Importing Modules in Python: Modules and Packages

```python
import math
from math import sqrt, pi 
math.sqrt(16)
```

```python

import numpy as np
np.array([1,2,3,4])
```

```python
from math import * 
print(sqrt(16))
print(pi)
```

```python
from package.maths import addition
addition(5,6)
```
