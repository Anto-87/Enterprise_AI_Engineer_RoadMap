#### Lambda functions in Python

```python
#syntax 

lambda arguments: expression
```

```python
addition = lambda a,b : f"The result : {a+b} "
type(addition)
```

```python
print(addition(5,6))
```

```python
even = lambda a : a%2==0
print(even(12))
```

```python
print(even(11))
```

```python
numbers=[1,2,3,4,5]
list(map(lambda x:x**2,numbers))
```
