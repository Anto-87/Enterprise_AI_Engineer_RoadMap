### Introduction to Functions

```python
## Syntax 

def function_name(parameters):
    """ Document string """
    return expression
```

```python
# Function definition 

def even_or_add(num):
    """ This function find the provided number as even or odd """
    if num%2==0:
        print("The given number is even")
    else:
        print("The given number is odd")
```

```python
# Call the function 

even_or_add(5)
```

```python
# function with multiple params 

def add(a,b):
    """ The function add the number provided in the arguments """
    return a+b
```

```python
add(5,6)
```

```python
def greet(name="Guest_User"):
    print(f"Welcome {name}!")

greet("Anto")
```

```python
greet() # without param, takes the default param value from the fn arguments
```

```python
### Variable length Argumnents 
### Positional and Keywords arguments

def print_numbers(*args):
    for number in args:
        print(number)
```

```python
print_numbers(1,2,3,4,5,6,"Anto",False)
```

```python
# Keywords Arguments
def print_details(** kwargs):
    for key, value in kwargs.items():
        print(f"{key}:{value}")
```

```python
print_details(name="krish", age="26", country="India")
```

```python
def print_details(*args, ** kwargs):
    for val in args:
        print(f"{val}")
    for key, value in kwargs.items():
        print(f"{key}:{value}")
```

```python
print_details(1,2,3,4,5,name="krish", age="26", country="India")
```
