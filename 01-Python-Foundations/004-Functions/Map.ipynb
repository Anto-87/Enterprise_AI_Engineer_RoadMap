#### The map() function in Python

```python
numbers=[1,2,3,4,5]
list(map(lambda x:x**2,numbers))
```

```python
numbers1=[1,2,3]
numbers2=[4,5,6]
added_numbers = list(map(lambda x, y:x+y, numbers1, numbers2))
```

```python
print(added_numbers)
```

```python
 words=['apple','cherry','banana']
 to_upper=list(map(str.upper, words))
 print(to_upper)
```

```python
def get_name(person):
    print(f"The person name is {person['name']}")
```

```python
persons=[
    {'name':"Anto", "age":34},
    {'name':'Chris',"age":39}
]

list(map(get_name, persons))
```
