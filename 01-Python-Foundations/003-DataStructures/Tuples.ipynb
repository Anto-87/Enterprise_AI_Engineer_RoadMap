```python
print(tuple_.count(4))
print(tuple_.index(1))
```

```python
#packing tuple
packed_tuple = 1,2,3,"hello",False
print(packed_tuple)
```

```python
#unpacking tuple 

a,b,c,d,e = packed_tuple 
print(e)

# unpacking with *
numbers =(1,2,3,4,5,6,7)
first,*middle,last=numbers
print(middle)
```

```python
list=[[1,2],[3,4],[True,"HAI"]]
print(list[0])
print(list[0][1])
```

```python
nested_tuple=((1,2,3),("Anto",False),(4,5,6))
print(nested_tuple[0])
print(nested_tuple[0][1])
```

```python
for tub in nested_tuple:
    for val in tub:
        print(val)
```
