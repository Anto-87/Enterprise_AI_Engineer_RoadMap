### Dictionaries in Python

```python
# Store the data in keyed/valued in pair.
empty_dic=dict()
print(type(empty_dic))
```

```python
student={"name":"Anto","age":34,"grade":24}
print(student)
print(type(student))
```

```python
student={"name":"Anto", "age":34,"name":45}
print(student)
```

```python
## Accessing the dictionaries elements

print(student['age'])
print(student.get('age'))
```

```python
# modify the dictionaries elements

student['age']=39
student['address']="India"
student['grade']=123
print(student)

del student['grade']
print(student)
```

```python
# Dictionaries methods 

keys=student.keys()
values =student.values()
print(keys)
print(values)

items = student.items()
print(items)
```

```python
## Shallow copy 
student
student_copy =student
print(student)
print(student_copy)
```

```python
student["name"]="Anto"
print(student)
print(student_copy)
```

```python
# To avoid the above scenario, not to update in copy variables

student_copy1=student.copy()
print(student_copy1)
```

```python
student["name"]="AntoChris"
print(student)
print(student_copy1)
```

```python
# iterating over dictionaries

for keys in student.keys():
    print(keys)
```

```python
for values in student.values():
    print(values)
    
```

```python
for key, value in student.items():
    print(f"{key}:{value}")
```

```python
# Nested dictionaries

students ={
    "student1":{"name":"Anto", "age":39},
    "student2":{"name":"Chris","age":43}
}

print(students)
```

```python
## Access the nested dictionaries

print(students["student2"]["name"])
print(students["student1"]["name"])
```

```python
for student_id,student_info in students.items():
    print(f"{student_id}:{student_info}")
    for key,value in student_info.items():
        print(f"{key}:{value}")
    
```

```python
# Dictionaries Comprehension
squares={x:x**2 for x in range(5)}
print(squares)
```

```python
squares={x:x**2 for x in range(5) if x%2==0}
print(squares)
```

```python
#merging the dictionaries

dict1= {"a":1,"b":2}
dict2={"b":3,"c":4}
merged_dict={**dict1,**dict2}
print(merged_dict)
```
