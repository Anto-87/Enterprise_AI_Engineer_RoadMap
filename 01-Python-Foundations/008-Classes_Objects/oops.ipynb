```python
class Dog:
    ## Constructor
    def __init__(self, name, age):
        self.name=name
        self.age=age
```

```python
dog2=Dog("Steffy",23)
print(dog2)
print(dog2.name)
print(dog2.age)
```

```python
## class wit instance methods
class Dog:
    def __init__(self,name,age):
        self.name=name
        self.age=age 
    
    def bark(self):
        print(f"{self.name} says woof")
    

dog = Dog('Buddy',13)
dog1 =Dog('Lucy',5)
dog.bark()
dog1.bark()
```
